```python
df.columns = df.columns.str.lower().str.replace(" ","_").str.replace("(","",regex = False).str.replace(")","",regex = False)
```

```python
columns_to_remove = ["customer_email",
    "customer_password",
    "category_id",
    "customer_state",
    "customer_zipcode",
    "department_id",
    "department_name",
    "order_zipcode",
    "product_card_id",
    "product_category_id",
    "product_description",
    "product_status"
]
df = df.drop(columns = columns_to_remove)
```
