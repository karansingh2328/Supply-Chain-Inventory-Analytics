```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/cleaned_data/DataCoSupplyChain_Cleaned.csv", parse_dates = ["order_date_dateorders","shipping_date_dateorders"])
```

```python
df.head()
```

```python
df.dtypes
```

```python
df["order_year"] = df["order_date_dateorders"].dt.year
df[["order_date_dateorders","order_year"]].head()
```

```python
df["order_month_name"] = df["order_date_dateorders"].dt.month_name()
df[["order_date_dateorders","order_month_name"]].head()
```

```python
df["order_quarter"] = df["order_date_dateorders"].dt.quarter
df[["order_date_dateorders","order_quarter"]].head()
```

```python
df["order_day_name"] = df["order_date_dateorders"].dt.day_name()
df[["order_date_dateorders","order_day_name"]].head()
```

```python
df["shipping_duration_days"] = (df["shipping_date_dateorders"]-df["order_date_dateorders"]).dt.days
df["shipping_duration_days"].head()
```

```python
df["is_weekend_order"] = df["order_day_name"].isin(["Saturday", "Sunday"])
df["is_weekend_order"].head()
```

```python
df["profit_margin_percentage"] = ((df["benefit_per_order"]/df["sales_per_customer"])*100).round(2)
df["profit_margin_percentage"].head()
```

```python
df["discount_amount"] = (df["order_item_quantity"] * df["order_item_discount"]).round(2)
df["discount_amount"].head()
```

```python
df["delivery_delay_days"] = df["days_for_shipping_real"] - df["days_for_shipment_scheduled"]
df["delivery_delay_days"].head()
```

```python
df["is_late_delivery"] = df["late_delivery_risk"] > 0
df["is_late_delivery"].head()
```

```python
df.to_csv(
    "../data/feature_engineered_data/dataco_feature_engineered.csv",
    index=False
)
```
